```python
# Function that returns a dictionary with person info, age is optional
def build_person(first_name, last_name, age=None):
    person = {'first': first_name, 'last': last_name}
    if age:
        person['age'] = age
    return person
```

## Passing a List

```python
# Function that takes a list of names and greets each one
def greet_users(names):
    """Print a simple greeting with given name in the list"""
    for name in names:
        msg = f"Hello, {name.title()}!"
        print(msg)
```

```python
# Initialize lists for unprinted and completed models
unprinted_designs = ["phone case", 'robot arm', "dodecahedron"]
completed_models = []

# Function to move items from unprinted to completed list
def print_model(list1, list2):
    while list1:
        current_model = list1.pop()
        print(f"Printing Model: {current_model}")
        list2.append(current_model)
```

```python
# Function to display completed models
def show_completed_models(models_list):
    print("The following models were printed:")
    for model in models_list:
        print(model)
```

```python
# Call function to process all unprinted designs
print_model(unprinted_designs, completed_models)
```

```python
# Display the completed models
show_completed_models(completed_models)
```

## Passing an Arbitrary Number of Arguments

```python
# Function with arbitrary number of positional arguments (*toppings becomes a tuple)
def make_pizza(size, *toppings):
    """Print the list of toppings that have been requested"""
    # This will return a tuple.
    print(f"\nMaking a pizza {size} inches with the following toppings:")
    for topping in toppings:
        print(f"-{topping}")

# Call function with 2 toppings
make_pizza(14, 'pepper', 'mushrooms')
# Call function with 3 toppings
make_pizza(12, 'pepperoni', 'mushrooms', 'tomato')
```

## Sandwiches: Write a function that accepts a list of items a person wants on a
sandwich. The function should have one parameter that collects as many items as the
function call provides, and it should print a summary of the sandwich that's being
ordered. Call the function three times, using a different number of arguments each
time.


```python
# Exercise: Function accepting arbitrary number of sandwich ingredients
def sandwich(*args):
    print("The Sandwich is made up following ingredients:")
    for _ in args:
        print(f"-{_}")
```

```python
# Call with no ingredients
sandwich()
# Call with 2 ingredients
sandwich("Tomato", "chees")
# Call with 2 different ingredients
sandwich("butter", "chees")
```

## Cars: Write a function that stores information about a car in a dictionary. The
function should always receive a manufacturer and a model name. It should then accept an arbitrary number of keyword arguments. Call the function with the required
information and two other name-value pairs, such as a color or an optional feature.
Your function should work for a call like this one:

car = make_car('Subaru', 'outback', color='blue', tow_package=True)

Print the dictionary that's returned to make sure all the information was stored
correctly.

```python
# Function to store car info with required params and arbitrary keyword args
def make_car(model_name, model_info, **car_info):
    """"Make a car model"""
    car_info["model"] = model_name
    car_info["model_info"] = model_info
    return car_info
```

```python
# Create car dictionary with required params and optional features
car = make_car('Subaru', 'outback', color='blue', tow_package=True)
```

```python
# Print the car dictionary to verify all info was stored
print(car)
```
