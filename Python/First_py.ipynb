## Return Value

```python
def get_formated_name(first_name, last_name, middle_name=""):
    """Return a formatted name"""
    full_name = f"{first_name} {middle_name} {last_name}"
    return full_name.title()
```

```python
musician = get_formated_name("Jimi", "log")
musician1 = get_formated_name("Jimi", "log", middle_name="Jimi")
```

```python
print(musician)
print(musician1)
```

```python
def build_person(first_name, last_name, age=None):
    person = {'first': first_name, 'last': last_name}
    if age:
        person['age'] = age
    return person
```

```python
person1 = build_person("Abhijeet", "giri", age=20)
print(person1)
```

```python
person2 = build_person("jimi", "log")
print(person2)
```
