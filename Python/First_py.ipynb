```python
def greet_user(username):
    "Display a simple greeting"
    print(F"Hello, {username.title()}!")
```

```python
def favorite_book(title):
    print("My favorite book is " + title.title())
```

```python
favorite_book("SDF")
```

### Multiple Arguments

```python
def describe_pet(pet_name, animal_type = "Dog"): # Default values
    print(f"I Have a {animal_type}.")
    print(f"My {animal_type} name is {pet_name.title()}.")
```

```python
describe_pet("Austin")
```

```python
# keyword arguments
describe_pet(pet_name="monty", animal_type="cat")
```
